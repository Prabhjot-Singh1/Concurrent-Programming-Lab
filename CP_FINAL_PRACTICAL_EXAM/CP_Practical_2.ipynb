```python
from datasets import load_dataset

# Fixed: use parquet format directly, no trust_remote_code needed
dataset = load_dataset("wikimedia/wikipedia", "20231101.en", split="train[:500]")

# Combine all articles into one big text
full_text = " ".join([item['text'] for item in dataset])
full_text = full_text.lower()

print(f"Loaded {len(dataset)} Wikipedia articles")
print(f"Total characters: {len(full_text)}")
print(f"\nSample text:\n{full_text[:300]}")
```

```python
import numpy as np
from numba import cuda

# Convert text to ASCII numbers (GPU can only work with numbers, not strings)
text_array = np.frombuffer(full_text.encode('ascii', errors='replace'), dtype=np.uint8)

print(f"Text converted to {len(text_array)} ASCII values")
print(f"Sample ASCII values: {text_array[:10]}")
print(f"Which means: {[chr(c) for c in text_array[:10]]}")
```

```python
# ── CUDA Kernel: count frequency of each letter ────────────────
@cuda.jit
def count_letters(text, letter_counts):
    idx = cuda.grid(1)   # each thread handles one character
    if idx < text.size:
        c = text[idx]
        if 97 <= c <= 122:   # ASCII 97='a', 122='z'
            cuda.atomic.add(letter_counts, c - 97, 1)  # add to that letter's slot

# ── Setup ───────────────────────────────────────────────────────
BLOCK_SIZE = 256
num_blocks = (text_array.size + BLOCK_SIZE - 1) // BLOCK_SIZE

# Copy to GPU
d_text         = cuda.to_device(text_array)
d_letter_counts = cuda.to_device(np.zeros(26, dtype=np.int32))  # 26 letters

# Launch kernel
count_letters[num_blocks, BLOCK_SIZE](d_text, d_letter_counts)

# Copy back to CPU
h_letter_counts = d_letter_counts.copy_to_host()

# Show results
import matplotlib.pyplot as plt

letters = [chr(i + 97) for i in range(26)]
print("\nLetter Frequencies:")
for i, letter in enumerate(letters):
    print(f"  '{letter}': {h_letter_counts[i]:,}")
```

```python
# ── CUDA Kernel: count spaces to estimate word count ───────────
@cuda.jit
def count_spaces(text, space_count):
    idx = cuda.grid(1)
    if idx < text.size:
        if text[idx] == 32:   # ASCII 32 = space
            cuda.atomic.add(space_count, 0, 1)

# Setup
d_space_count = cuda.to_device(np.zeros(1, dtype=np.int32))

# Launch
count_spaces[num_blocks, BLOCK_SIZE](d_text, d_space_count)

# Result
total_words = d_space_count.copy_to_host()[0] + 1
print(f"Estimated total words: {total_words:,}")
```

```python
# ── CUDA Kernel: check if a window of characters matches keyword ─
@cuda.jit
def count_keyword(text, keyword, result, klen):
    idx = cuda.grid(1)
    if idx <= text.size - klen:
        match = True
        for i in range(klen):
            if text[idx + i] != keyword[i]:
                match = False
                break
        if match:
            cuda.atomic.add(result, 0, 1)

# ── Keywords to search ──────────────────────────────────────────
keywords = ["science", "history", "war", "language", "art",
            "music", "physics", "biology", "computer", "india"]

print("Keyword Counts in Wikipedia Corpus:")
print("-" * 35)

keyword_results = {}
for word in keywords:
    kw_array  = np.frombuffer(word.encode('ascii'), dtype=np.uint8)
    d_keyword = cuda.to_device(kw_array)
    d_result  = cuda.to_device(np.zeros(1, dtype=np.int32))

    count_keyword[num_blocks, BLOCK_SIZE](d_text, d_keyword, d_result, len(word))

    count = d_result.copy_to_host()[0]
    keyword_results[word] = count
    print(f"  '{word}': {count:,} times")
```

```python
fig, axes = plt.subplots(1, 3, figsize=(18, 5))

# Plot 1: Letter frequency
axes[0].bar(letters, h_letter_counts, color='steelblue')
axes[0].set_title("Letter Frequency (CUDA)", fontsize=13, fontweight='bold')
axes[0].set_xlabel("Letter")
axes[0].set_ylabel("Count")

# Plot 2: Top 10 letters
top_indices = np.argsort(h_letter_counts)[::-1][:10]
top_letters = [letters[i] for i in top_indices]
top_counts  = [h_letter_counts[i] for i in top_indices]
axes[1].bar(top_letters, top_counts, color='coral')
axes[1].set_title("Top 10 Most Frequent Letters", fontsize=13, fontweight='bold')
axes[1].set_xlabel("Letter")
axes[1].set_ylabel("Count")

# Plot 3: Keyword counts
axes[2].bar(keyword_results.keys(), keyword_results.values(), color='mediumseagreen')
axes[2].set_title("Keyword Counts (CUDA)", fontsize=13, fontweight='bold')
axes[2].set_xlabel("Keyword")
axes[2].set_ylabel("Count")
axes[2].tick_params(axis='x', rotation=45)

plt.suptitle("CUDA Text Analytics — Wikipedia Corpus (DS11)",
             fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()

print(f"\nSummary:")
print(f"  Total characters : {len(text_array):,}")
print(f"  Total words      : {total_words:,}")
print(f"  Most common letter: '{letters[np.argmax(h_letter_counts)]}'")
print(f"  Most common keyword: '{max(keyword_results, key=keyword_results.get)}'")
```
